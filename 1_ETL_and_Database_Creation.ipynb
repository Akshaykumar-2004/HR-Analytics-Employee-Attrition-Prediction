```python
# === Step 1: Import Your Toolkits ===
# What are we doing? Gathering the tools we need.
# pandas: For working with data tables (DataFrames).
# sqlalchemy: Lets Python speak the language of databases.
# logging: Our script's personal diary to record what it does.
# time: A stopwatch to see how long our process takes.
import pandas as pd
from sqlalchemy import create_engine
import logging
import time

# === Step 2: Set Up the Script's Diary (Logging) ===
# What are we doing? Telling our script to create a log file in our project folder.
# Why? So we can check that everything worked correctly. It's a professional habit.
log_path = "hr_project_ingestion.log"
logging.basicConfig(
    filename=log_path,
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filemode='w'  # 'w' starts a fresh diary every time we run the script.
)

# === Step 3: Create the "Key" to Our Database ===
# What are we doing? Defining where our database will live and its name.
# Why? All our organized data will be in one file, making it easy to access.
db_path = "hr_analytics.db"
engine = create_engine('sqlite:///' + db_path)
logging.info("Database engine created. It will be saved at: " + db_path)

# === Step 4: The Master Function to Run the Whole Process ===
# What are we doing? Putting all instructions into one big function called 'load_data_to_db'.
# Why? Organizing code into functions makes it clean and reusable.
def load_data_to_db():
    """Reads the HR CSV file and loads it into our SQLite database."""
    start_time = time.time()  # Start the stopwatch.
    logging.info("--- Starting ETL Process ---")

    try:  # 'try' means: "Attempt this, but be ready for errors."
        # Define the full path to our evidence file.
        csv_file_path = 'data/WA_Fn-UseC_-HR-Employee-Attrition.csv'
        
        # Define a clean, simple name for our table in the database.
        table_name = 'hr_data'
        logging.info("Processing file: " + csv_file_path + " -> into table: " + table_name)

        # Use pandas to read the entire CSV file into memory.
        df = pd.read_csv(csv_file_path)

        # This is the magic 'Load' step.
        # df.to_sql() saves our pandas DataFrame as a table inside our database file.
        # name=table_name: The name of the table in the database.
        # con=engine: The database connection we want to use.
        # if_exists='replace': If the table already exists, delete it and create a new one.
        # index=False: We don't need to save the pandas index number as a column.
        df.to_sql(name=table_name, con=engine, if_exists='replace', index=False)
        logging.info("Successfully loaded table: " + table_name)

    except Exception as e:  # If any error happens...
        # ...we 'catch' it and record it in our diary instead of crashing.
        logging.error("Failed to process file. Error: " + str(e))

    end_time = time.time()  # Stop the stopwatch.
    total_time_seconds = end_time - start_time
    logging.info("--- ETL Process Complete. Total time taken: " + str(round(total_time_seconds, 2)) + " seconds. ---")

# === Step 5: The "Start Button" for our Script ===
# What is this? A standard Python safety check.
# It means: "Only run the 'load_data_to_db' recipe if this script is being run directly."
if __name__ == '__main__':
    load_data_to_db()
    print("ETL Process has finished. Check the log file 'hr_project_ingestion.log' for details.")
```
